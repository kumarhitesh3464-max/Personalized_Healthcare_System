# Personalized Healthcare & Medicine Recommendation System

This notebook performs data preprocessing and prepares the healthcare data for machine learning.

```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

## 1. Load Healthcare Datasets

This section loads all datasets required for preprocessing and model development.

```python
# Load main datasets

df1 = pd.read_csv("../data/dataset_1/dataset.csv")
df2 = pd.read_csv("../data/dataset_2/medical data.csv")
df3 = pd.read_csv("../data/dataset_3/Diseases_and_Symptoms_dataset.csv")

# Load supporting datasets

df1_severity = pd.read_csv("../data/dataset_1/Symptom-severity.csv")
df1_description = pd.read_csv("../data/dataset_1/symptom_Description.csv")
df1_precaution = pd.read_csv("../data/dataset_1/symptom_precaution.csv")

df3_description = pd.read_csv("../data/dataset_3/description (2).csv")
df3_diets = pd.read_csv("../data/dataset_3/diets.csv")
df3_medications = pd.read_csv("../data/dataset_3/medications.csv")
df3_precautions = pd.read_csv("../data/dataset_3/precautions.csv")
df3_workout = pd.read_csv("../data/dataset_3/workout.csv")

print("All datasets loaded successfully!")
```

```python
print("Dataset 1:", df1.shape)
print("Dataset 2:", df2.shape)
print("Dataset 3:", df3.shape)

print("\nSupporting datasets:")
print("Severity:", df1_severity.shape)
print("Description:", df1_description.shape)
print("Precaution:", df1_precaution.shape)
print("Diets:", df3_diets.shape)
print("Medications:", df3_medications.shape)
print("Precautions:", df3_precautions.shape)
print("Workout:", df3_workout.shape)
```

## 2. Symptom Standardization

This section cleans and standardizes symptom names to maintain consistency across the datasets.

```python
# Standardize column names and symptom values

def clean_text(value):
    if pd.isna(value):
        return value
    
    value = str(value).strip().lower()
    value = value.replace("_", " ")
    value = " ".join(value.split())
    
    return value


# Clean Dataset 1 symptom columns
symptom_cols_df1 = [
    col for col in df1.columns
    if col.lower().startswith("symptom")
]

for col in symptom_cols_df1:
    df1[col] = df1[col].apply(clean_text)


# Clean Dataset 2 symptoms
df2["Symptoms"] = df2["Symptoms"].apply(clean_text)


# Clean Dataset 3 symptom columns
symptom_cols_df3 = [
    col for col in df3.columns
    if "symptom" in col.lower()
]

for col in symptom_cols_df3:
    df3[col] = df3[col].apply(clean_text)


print("Symptom standardization completed successfully!")
print("\nDataset 1 symptom columns:", len(symptom_cols_df1))
print("Dataset 3 symptom columns:", len(symptom_cols_df3))
```

```python
# Display cleaned symptom data

print("Dataset 1:")
display(df1[symptom_cols_df1].head())

print("\nDataset 2:")
display(df2[["Symptoms"]].head())

print("\nDataset 3:")
display(df3[symptom_cols_df3].head())
```

## 3. Disease Label Standardization

This section cleans and standardizes disease names for consistent machine learning labels.

```python
# Standardize disease names

# Dataset 1
df1["Disease"] = df1["Disease"].apply(clean_text)

# Dataset 2
df2["Disease"] = df2["Disease"].apply(clean_text)

# Dataset 3
# Detect disease column automatically
possible_disease_columns = [
    col for col in df3.columns
    if any(word in col.lower() for word in ["disease", "diagnosis", "prognosis"])
]

if len(possible_disease_columns) == 0:
    raise ValueError("Disease column not found in Dataset 3.")

disease_col_df3 = possible_disease_columns[0]

df3[disease_col_df3] = df3[disease_col_df3].apply(clean_text)

# Clean text columns in supporting datasets
supporting_datasets = [
    df1_description,
    df1_precaution,
    df3_description,
    df3_diets,
    df3_medications,
    df3_precautions,
    df3_workout
]

for data in supporting_datasets:
    for col in data.select_dtypes(include="object").columns:
        data[col] = data[col].apply(clean_text)

print("Disease label standardization completed successfully!")
print("Dataset 3 disease column:", disease_col_df3)
```

```python
# Verify standardized disease labels

print("Dataset 1 Diseases:")
display(df1["Disease"].value_counts().head(10))

print("\nDataset 2 Diseases:")
display(df2["Disease"].value_counts().head(10))

print("\nDataset 3 Diseases:")
display(df3[disease_col_df3].value_counts().head(10))
```

## 4. Symptom Feature Engineering

This section converts symptoms into binary features for machine learning.

```python
# Create a clean symptom list from Dataset 1

all_symptoms = set()

for col in symptom_cols_df1:
    for value in df1[col].dropna():
        value = str(value).strip()
        
        if value and value.lower() != "nan":
            all_symptoms.add(value)

all_symptoms = sorted(all_symptoms)

print("Total Unique Symptoms:", len(all_symptoms))
print("\nFirst 20 Symptoms:")
print(all_symptoms[:20])
```

```python
# Create binary symptom features

df1_ml = pd.DataFrame(0, index=df1.index, columns=all_symptoms)

for col in symptom_cols_df1:
    for idx, value in df1[col].items():
        if pd.notna(value):
            value = str(value).strip()
            
            if value in df1_ml.columns:
                df1_ml.loc[idx, value] = 1

# Add disease target
df1_ml["Disease"] = df1["Disease"]

print("ML-ready Dataset Shape:", df1_ml.shape)

display(df1_ml.head())
```

```python
# Verify binary symptom features

print("Feature values:")
print(df1_ml[all_symptoms].stack().value_counts())

print("\nDisease classes:", df1_ml["Disease"].nunique())
print("Total records:", len(df1_ml))
```

## 5. ML Feature Quality Check

This section validates the engineered symptom features before model training.

```python
# Validate ML-ready feature matrix

X_features = df1_ml.drop(columns=["Disease"])
y_target = df1_ml["Disease"]

print("=" * 70)
print("ML FEATURE QUALITY CHECK")
print("=" * 70)

print("Feature Matrix Shape:", X_features.shape)
print("Target Shape:", y_target.shape)

print("\nMissing Values in Features:", X_features.isnull().sum().sum())
print("Duplicate Feature Rows:", X_features.duplicated().sum())

print("\nFeature Value Distribution:")
print(X_features.stack().value_counts())

print("\nNumber of Disease Classes:", y_target.nunique())

print("\nClass Distribution:")
display(y_target.value_counts().head(15))
```

## 6. Feature and Target Separation

The symptom features are separated from the disease target to prepare the data for model training.

```python
# Final feature and target preparation

X = df1_ml.drop(columns=["Disease"]).copy()
y = df1_ml["Disease"].copy()

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nTotal Features:", X.shape[1])
print("Total Disease Classes:", y.nunique())

print("\nFeatures are binary:", set(X.values.flatten()).issubset({0, 1}))
```

## 7. Stratified Train-Test Split

The dataset is divided into training and testing sets using stratification to preserve disease-class distribution.

```python
from sklearn.model_selection import train_test_split

# Stratified train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("=" * 70)
print("TRAIN-TEST SPLIT")
print("=" * 70)

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("Training Target:", y_train.shape)
print("Testing Target:", y_test.shape)

print("\nTraining Disease Classes:", y_train.nunique())
print("Testing Disease Classes:", y_test.nunique())
```

## 8. Verify Stratification

This section verifies that disease-class proportions are preserved between the training and testing datasets.

```python
# Compare disease distribution

train_distribution = y_train.value_counts(normalize=True).sort_index()
test_distribution = y_test.value_counts(normalize=True).sort_index()

distribution_comparison = pd.DataFrame({
    "Train (%)": train_distribution * 100,
    "Test (%)": test_distribution * 100
})

distribution_comparison["Difference (%)"] = (
    distribution_comparison["Train (%)"] -
    distribution_comparison["Test (%)"]
).abs()

display(distribution_comparison)

print("\nMaximum distribution difference:",
      round(distribution_comparison["Difference (%)"].max(), 3), "%")
```

## 9. Reproducibility Configuration

A fixed random seed is used to ensure consistent and reproducible machine learning experiments.

```python
# Reproducibility
RANDOM_STATE = 42

print("Random State:", RANDOM_STATE)
print("Train-Test Split: 80%-20%")
print("Stratification: Enabled")
```

## 10. Disease Label Encoding

Disease names are encoded into numerical labels for machine learning while preserving the original disease names for prediction output.

```python
from sklearn.preprocessing import LabelEncoder

# Initialize label encoder
label_encoder = LabelEncoder()

# Fit only on the target labels
y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)

print("=" * 70)
print("DISEASE LABEL ENCODING")
print("=" * 70)

print("Number of Disease Classes:", len(label_encoder.classes_))

print("\nEncoded Labels:")
for encoded_label, disease in enumerate(label_encoder.classes_):
    print(f"{encoded_label:2d} → {disease}")

print("\nTraining Encoded Shape:", y_train_encoded.shape)
print("Testing Encoded Shape:", y_test_encoded.shape)
```

```python
# Verify that encoding and decoding work correctly

sample_encoded = y_train_encoded[:10]
sample_decoded = label_encoder.inverse_transform(sample_encoded)

print("Encoded values:")
print(sample_encoded)

print("\nDecoded disease names:")
print(sample_decoded)
```

## 11. Final Preprocessed Dataset Check

This section performs a final validation before machine learning model development.

```python
# Final preprocessing validation

print("=" * 70)
print("FINAL PREPROCESSING VALIDATION")
print("=" * 70)

checks = {
    "Training samples": len(X_train),
    "Testing samples": len(X_test),
    "Features": X_train.shape[1],
    "Disease classes": len(label_encoder.classes_),
    "Training missing values": X_train.isnull().sum().sum(),
    "Testing missing values": X_test.isnull().sum().sum(),
    "Training duplicate rows": X_train.duplicated().sum(),
    "Testing duplicate rows": X_test.duplicated().sum(),
}

for check, value in checks.items():
    print(f"{check}: {value}")

print("\nBinary features:",
      set(X_train.values.flatten()).issubset({0, 1}))

print("\nPreprocessing validation completed.")
```
