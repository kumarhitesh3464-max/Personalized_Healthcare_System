## 6. Disease Distribution Analysis

This section analyzes the distribution of diseases across the healthcare datasets.

```python
# Disease Distribution Analysis

print("=" * 70)
print("DISEASE DISTRIBUTION ANALYSIS")
print("=" * 70)

# Display Dataset 1 columns
print("\nDataset 1 Columns:")
print(df1.columns.tolist())

# Possible disease column names
possible_disease_columns = [
    "Disease",
    "disease",
    "Disease Name",
    "disease_name",
    "prognosis",
    "Prognosis",
    "diagnosis",
    "Diagnosis"
]

# Find disease column automatically
disease_column = None

for column in possible_disease_columns:
    if column in df1.columns:
        disease_column = column
        break

# If disease column is found
if disease_column is not None:
    
    disease_counts = df1[disease_column].value_counts()

    print("\nDisease Column:", disease_column)
    print("Total Unique Diseases:", disease_counts.shape[0])

    print("\nTop 10 Diseases:")
    display(disease_counts.head(10))

else:
    print("\nDisease column was not found automatically.")
    print("Please check the Dataset 1 column names shown above.")
```

## 7. Disease Distribution Visualization

This section visualizes the distribution of diseases in Dataset 1.

```python
# Disease Distribution Bar Chart

plt.figure(figsize=(14, 6))

df1["Disease"].value_counts().plot(kind="bar")

plt.title("Disease Distribution in Dataset 1")
plt.xlabel("Disease")
plt.ylabel("Number of Records")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

## 8. Symptom Frequency Analysis

This section analyzes the frequency of symptoms in Dataset 1.

```python
# Symptom Frequency Analysis

symptom_columns = [
    col for col in df1.columns
    if col.lower().startswith("symptom")
]

print("Symptom Columns:")
print(symptom_columns)

# Count symptom occurrences
symptom_counts = {}

for column in symptom_columns:
    for symptom in df1[column].dropna():
        symptom = str(symptom).strip()
        
        if symptom and symptom.lower() != "nan":
            symptom_counts[symptom] = symptom_counts.get(symptom, 0) + 1

# Convert to Series
symptom_counts = pd.Series(symptom_counts).sort_values(ascending=False)

print("\nTotal Unique Symptoms:", len(symptom_counts))

print("\nTop 15 Most Frequent Symptoms:")
display(symptom_counts.head(15))
```

## 9. Symptom Frequency Visualization

This section visualizes the most frequent symptoms found in Dataset 1.

```python
# Top 15 Symptoms Visualization

top_symptoms = symptom_counts.head(15)

plt.figure(figsize=(12, 6))

top_symptoms.sort_values().plot(kind="barh")

plt.title("Top 15 Most Frequent Symptoms")
plt.xlabel("Number of Records")
plt.ylabel("Symptoms")
plt.tight_layout()
plt.show()
```

## 14. Symptom Frequency Analysis in Dataset 2

This section analyzes the most frequently reported symptoms among patients.

```python
# Symptom Frequency Analysis - Dataset 2

symptom_counts_df2 = {}

for symptoms in df2["Symptoms"].dropna():
    symptoms = str(symptoms).strip()
    
    # Split symptoms using comma
    for symptom in symptoms.split(","):
        symptom = symptom.strip()
        
        if symptom:
            symptom_counts_df2[symptom] = symptom_counts_df2.get(symptom, 0) + 1

# Convert to Series and sort
symptom_counts_df2 = pd.Series(symptom_counts_df2).sort_values(ascending=False)

print("Total Unique Symptoms:", len(symptom_counts_df2))

print("\nTop 15 Most Frequently Reported Symptoms:")
display(symptom_counts_df2.head(15))
```

## 19. Dataset 3 Disease Analysis

This section identifies the disease column and analyzes its distribution.

```python
# Dataset 3 Disease Column Detection

print("=" * 70)
print("DATASET 3 DISEASE ANALYSIS")
print("=" * 70)

# Find possible disease-related columns
possible_columns = [
    col for col in df3.columns
    if any(word in col.lower() for word in ["disease", "diagnosis", "prognosis"])
]

print("\nPossible Disease Columns:")
print(possible_columns)

# Automatically select disease column
if len(possible_columns) > 0:
    disease_col_df3 = possible_columns[0]
    
    print("\nSelected Disease Column:", disease_col_df3)
    print("Total Unique Diseases:", df3[disease_col_df3].nunique())
    
    print("\nTop 15 Diseases:")
    display(df3[disease_col_df3].value_counts().head(15))

else:
    print("\nNo disease-related column found.")
    print("Please check the column names printed above.")
```

## 21. Dataset 3 Symptom Analysis

This section identifies symptom-related columns and analyzes their frequency in Dataset 3.

```python
# Dataset 3 Symptom Column Analysis

print("=" * 70)
print("DATASET 3 SYMPTOM ANALYSIS")
print("=" * 70)

# Identify symptom-related columns
symptom_columns_df3 = [
    col for col in df3.columns
    if "symptom" in col.lower()
]

print("\nTotal Symptom Columns:", len(symptom_columns_df3))

print("\nSymptom Columns:")
print(symptom_columns_df3)

# Count symptom occurrences
symptom_frequency_df3 = {}

for column in symptom_columns_df3:
    for value in df3[column].dropna():
        value = str(value).strip()
        
        if value and value.lower() not in ["nan", "none", "no"]:
            symptom_frequency_df3[value] = (
                symptom_frequency_df3.get(value, 0) + 1
            )

# Convert to Series
symptom_frequency_df3 = pd.Series(
    symptom_frequency_df3
).sort_values(ascending=False)

print("\nTotal Unique Symptoms:", len(symptom_frequency_df3))

print("\nTop 20 Most Frequent Symptoms:")
display(symptom_frequency_df3.head(20))
```

## 22. Dataset 3 Symptom Frequency Visualization

This section visualizes the most frequently occurring symptoms in Dataset 3.

```python
# Top 20 Symptoms Visualization - Dataset 3

top_symptoms_df3 = symptom_frequency_df3.head(20)

plt.figure(figsize=(12, 7))

top_symptoms_df3.sort_values().plot(kind="barh")

plt.title("Top 20 Most Frequent Symptoms in Dataset 3")
plt.xlabel("Number of Records")
plt.ylabel("Symptoms")
plt.tight_layout()
plt.show()
```
